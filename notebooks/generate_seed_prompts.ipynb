Prompts generated using GPT-OSS-120B. Note that descriptions of the tools are appended to the individual prompts, inside of run_experiments.py.

```python
from promptolution.llms import APILLM
import yaml
```

```python
llm = APILLM(
    model_id="gpt-oss-120b-llmlb",
    api_url="https://openwebui.uni-freiburg.de/api",
    api_key=open("../token.txt").read().strip(),
)
```

```python
SETUP_YAML = "../configs/setups/mbpp.yaml"
TEMPLATE_PATH = "../configs/prompt_creation.txt"
```

```python
with open(SETUP_YAML) as f:
    setup = yaml.safe_load(f)

with open(TEMPLATE_PATH) as f:
    template = f.read()

for agent in setup["agents"]:
    print(agent["name"])
```

```python
agent = setup["agents"][1]

name = agent["name"]
task_desc = agent["task_description"]

inputs = ", ".join(f"{{{v}}}" for v in agent["input_vars"])
outputs = ", ".join(agent["output_vars"])

filled = (
    template.replace("<task_description>", task_desc)
    .replace("<Inputs>", inputs)
    .replace("<Outputs>", outputs)
)
prompts = llm.get_response(filled)
print(prompts[0])
```
